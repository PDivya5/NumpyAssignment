A2. Explain Vectorization
Definition

Vectorization is the process of performing an operation on an entire NumPy array at once instead of using an explicit Python loop.

Answer

Vectorization makes numerical computations faster because NumPy performs many operations internally using optimized code. It reduces the need for Python-level loops and becomes especially useful when working with large datasets.

```python
#NORMAL LOOP
result = []
for x in [1, 2, 3]:
    result.append(x * 2)
```

```python
#VECTORIZED NUMPY CODE
import numpy as np

a = np.array([1, 2, 3])
result = a * 2
print(result)
```

One-line explanation

Vectorization performs operations on the complete array efficiently without explicit Python loops.

```python
#EXAMPLE
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print(A * B)
print(A @ B)
```

A @ B performs matrix multiplication.


Neural-network examples


A * B: Can be used for element-wise operations such as applying masks or element-wise activation-related calculations.


A @ B: Commonly used for calculating the weighted sum in a neural network:


X @ W + b

A7. Normalization vs Standardization

Definition: Normalization

Normalization scales values into a fixed range, commonly 0 to 1.

Formula:

x' = (x - min) / (max - min)

Definition: Standardization

Standardization converts values so that they generally have:

Mean = 0
Standard deviation = 1

Formula:

z = (x - mean) / standard deviation

Why are they important?

Features may have very different scales. For example:

Age:       18–60
Salary:    20,000–500,000

Large-scale features may dominate certain ML algorithms.


```python
#NumPy example
import numpy as np

x = np.array([10, 20, 30, 40, 50])

normalized = (x - x.min()) / (x.max() - x.min())
standardized = (x - x.mean()) / x.std()

print(normalized)
print(standardized)
```

C5. Calculate y_pred and MSE

Definition

The linear model equation is:

y_pred = X @ w + b

MSE (Mean Squared Error) measures the average squared difference between actual and predicted values.

Formula:

MSE = mean((y_true - y_pred)²)

```python
import numpy as np

X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

w = np.array([0.5, 1.0])
b = 2.0

y_true = np.array([5, 7, 9])

y_pred = X @ w + b

mse = np.mean((y_true - y_pred) ** 2)

print("Predicted:", y_pred)
print("MSE:", mse)
```

Explanation

X @ w + b calculates predictions, and MSE calculates the average squared prediction error.

D1. Data Preprocessing Pipeline

You are given a raw 1D or 2D NumPy array containing NaN values.

Describe, step by step, the NumPy operations you would apply to:

(a) detect missing values,

(b) handle them appropriately and justify your strategy, and

(c) normalize or standardize the clean data before it is fed to a model.

Implement the full pipeline in code on a sample array of your own choosing, including at least one NaN.

Explanation

When a NumPy array contains NaN values, we can use the following steps:

Detect missing values using np.isnan().


Handle missing values using np.nanmean() to calculate the mean while ignoring NaN.

Replace the missing values with the calculated mean.

```python
import numpy as np

data = np.array([10, 20, np.nan, 40, 50])

# Detect missing values
missing = np.isnan(data)
print("Missing values:", missing)

# Calculate mean ignoring NaN
mean_value = np.nanmean(data)

# Replace NaN with mean
clean_data = np.where(missing, mean_value, data)
print("Clean data:", clean_data)

# Standardize the clean data
standardized = (clean_data - np.mean(clean_data)) / np.std(clean_data)
print("Standardized data:", standardized)
```

Explanation: np.isnan() detects missing values, np.nanmean() calculates the mean without considering NaN, and the missing value is replaced before standardization.

E1. Per-Column Min–Max Normalization 

Question

Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.


Answer


Min-max normalization scales each feature independently to the range 0 to 1. For each column, we subtract its minimum value and divide by the difference between its maximum and minimum values.

```python
import numpy as np

X = np.array([[10, 100],
              [20, 200],
              [30, 300],
              [40, 400]])

column_min = X.min(axis=0)
column_max = X.max(axis=0)

X_normalized = (X - column_min) / (column_max - column_min)

print("Original matrix:")
print(X)

print("Column-wise normalized matrix:")
print(X_normalized)
```

Explanation



axis=0 calculates the minimum and maximum separately for each column. Broadcasting then applies these values to every row, so each feature is normalized independently without using any explicit loops.

E2. Three-Level Score Classification 

Question

Given a 1D array of exam scores, use a single vectorized expression (no if/else statements and no explicit loops) to classify each score as:


Fail (< 40)

Average (40–70 inclusive)

Excellent (> 70)


Hint: You may need to nest or combine np.where() calls.


Answer



We can use nested np.where() to classify all scores at once. The first condition checks whether the score is below 40. If not, the second np.where() checks whether the score is 70 or below.

```python
import numpy as np

scores = np.array([25, 40, 55, 70, 85, 95])

result = np.where(scores < 40, "Fail",
                  np.where(scores <= 70, "Average", "Excellent"))

print("Scores:", scores)
print("Classification:", result)
```

Explanation


The first np.where() assigns Fail to scores below 40. For the remaining scores, the nested np.where() assigns Average to scores from 40 to 70 and Excellent to scores above 70. This performs the complete classification without loops or if/else statements.
